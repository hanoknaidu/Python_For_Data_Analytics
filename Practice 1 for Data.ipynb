## Break,Continue,Else,Pass

```python
# Break
i=0
while i <=10:
 
 if i ==6:
    break
 print(i)
 i +=1
```

```python
integers = [1,2,3,4,5,6]

for num in integers:
    power_of_num = num**num
    if power_of_num>3000:
        break
    print(power_of_num)
```

```python
#Continue
number=0
while number<5:
    number=number+1
    if number==3:
        continue
    print(number)
```

```python
integers=[1,2,3,4,5,6,7]
for num in integers:
    power_of_num = num**num
    if 200<=power_of_num<=3250:
        continue
    print(num,power_of_num)
```

```python
i=0
while i <=10:
 
 if i ==6:
    break
 print(i)
 i +=1
else:
    print("no breaks found")
```

```python
integers = [1,2,3,4,5,6]

for num in integers:
    power_of_num = num**num
    if power_of_num>3000:
        break
    print(power_of_num)
else:
    power_of_num=num**num
    print(power_of_num)
```

## Nested While Loops

```python
number = 0
while number <=10:
    print(number)
    inner_number=0
    while inner_number<5:
        print(inner_number)
        if inner_number==2:
            break
        inner_number +=1
    else:
      print('Inner while loop ended')
    number +=1
else:
       print('Outer while loop ended')
    
 
```

## List Comprehension

```python
flavors=['vanilla','chocolate','caramel','cinnamon','mango']
choco_flavors=[]
for flavor in flavors:
    if 'caramel' in flavor:
        choco_flavors.append(flavor)
print(choco_flavors)
```

```python
choco_flavors2=[flavor for flavor in flavors if 'caramel' in flavors]
print(choco_flavors2)
```

```python
new_list=[x**2 for x in range(10) if x%2==0]
print(new_list)
```
