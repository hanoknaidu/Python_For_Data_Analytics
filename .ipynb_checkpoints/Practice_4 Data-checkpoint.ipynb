## Try Except Basics

```python
x=10
y=20
```

```python
print(x)
```

```python
try:
    print(x)
except Exception as e: 
    print(e)
except NameError:
    print("Variable not Defined")
except:
     print("there is an error")
else:
    print("else was triggered")
finally:
    print("finnally")
```

## Getting A Text from Website

```python
from bs4 import BeautifulSoup
import requests
url='https://webscraper.io/'
page=requests.get(url)
soup=BeautifulSoup(page.text,'html')
```

```python
all_modal_title=soup.find_all('h4','modal-title')
```

```python
all_links=soup.find_all('a','v2-announcement__link')
```

```python
all_modals=[modals.text for modals in all_modal_title]
print(all_modals)
```

```python
# Split each block of text into separate lines/items
all_links = [line.strip() for text in all_links for line in text.text.split('\n') if line.strip()]
del all_links[2]
print(all_links)
```

```python
print(type(all_links))
```

```python
import pandas as pd
pd.DataFrame(all_modals,all_links,columns=['Modals'])
```
