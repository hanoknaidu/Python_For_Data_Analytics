## Regex MetaCharacters

```python
import re
```

```python
quote="my name is alex.My phone number is 91273874982.my email id is hanoknaidu123@gmail.com"
```

```python
#d - numbers
re.findall('\d',quote)
```

```python
#w - word character
re.findall('\w',quote)
```

```python
#W -oppostie of word character
re.findall('\W',quote)
```

```python
quote="my name is alex.My phone number is 91273874982.my email id is hanoknaidu123@gmail.com"
```

```python
#\a start of string
re.findall('^[a-z]',quote)
```

```python
#\a start of string
re.findall('\A[a-z]',quote)
```

```python
#\z end of string
re.findall('[a-z]\Z',quote)
```

```python
#\b, word boundary
string='I like python is pythonlove a word'
re.findall('\b',string)
```

```python
string='I like python is pythonlove a word'
re.findall(r'\bpython\b',string)
```

```python
# \B not a word boundary
string='I like python is pythonlove a word'
re.findall('\Bython\B',string)
```

```python
string='there are 365days in a year.8760 hours.5468098 minutes.Only 24 days until christmas.1 day at a time.'
re.findall(r'\d{3,4}',string)
```

```python
string='there are 365days in a year.8760 hours.5468098 minutes.Only 24 days until christmas.1 day at a time.'
re.findall(r'\d+',string)
```

```python
len(re.findall(r'\d+',string))
```

## Match Object Methods

```python
import re
```

```python
string="It's BEAUTIFUL day in the city.It's 75 degrees and sunny!"
```

```python
re.search(r"(\b[A-Z]+\b).*(\b\d+\b)",string)
```

```python
result=re.search(r"(\b[A-Z]+\b).*(\b\d+\b)",string)
```

```python
result.groups()
```

```python
result.group(1)
```

```python
result.group(2)
```

```python
result.span()
```

```python
result.start()
```

```python
result.end()
```

## Regex Use Cases

```python
import re

random_text= """
My name is Alex.My phone number is 9123456789.My E-mail id is alex123@gmail.com
My name is John.My phone number is 8136836477.My E-mail id is john234@yahoo.com
My name is Harry.My phone number is 712345689.My E-mail id is harry456@apple.com
"""
```

```python
re.findall(r"@[a-zA-Z]+",random_text)
```

```python
re.findall(r"@[\w]+",random_text)
```

```python
re.findall(r"[\w\.]+@[\w\.+]+",random_text)
```

```python
random_text= """
My name is Alex.My phone number is 912-345-6789.My E-mail id is alex123@gmail.com
My name is John.My phone number is 813-683-6477.My E-mail id is john234@yahoo.com
My name is Harry.My phone number is 712-345-6891.My E-mail id is harry456@apple.com
"""
```

```python
re.findall(r"\d{3}-\d{3}-\d{4}",random_text)
```

```python
re.findall(r"\d{10}",random_text)
```

```python
re.findall(r"\ (\d{3})",random_text)
```

```python
email_list=['alex123@gmail.com','john234@yahoo.com','harry456@apple.com']
```

```python
for email in email_list:
    result=re.findall(r"\@[a-zA-Z]+",email)
    print(result)
```

```python
domain_list=[re.findall(r"@[a-zA-z]+",email)[0]for email in email_list]
print(domain_list)
```
