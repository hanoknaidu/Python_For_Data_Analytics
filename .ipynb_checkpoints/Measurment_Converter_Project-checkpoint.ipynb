## Unit of Measurement Converter

```python
convert_from =input("Enter Starting Unit of Measurement(inches,feets,yards): ")

convert_to =input("Enter Unit of Measurement to Convert to(inches,feets,yards): ")

number_of_inches =input("Enter Starting Measurements in Inches:")

number_of_feets=input("Enter Starting Measurements in Feets:")

number_of_yards=input("Enter Starting Measurements in Yards:")
```

```python
convert_from =input("Enter Starting Unit of Measurement(inches,feets,yards): ")

convert_to =input("Enter Unit of Measurement to Convert to(inches,feets,yards): ")

if convert_from.lower() in ["inches",'in','inch']:
   number_of_inches =int(input("Enter Starting Measurements in Inches:"))
   if convert_to.lower() in ["feets",'ft','feet','foot']:
        print("Result:" + str(number_of_inches) +" Inches= " +str(round(number_of_inches/12,2)) +" Feet")
   elif convert_to.lower() in ["yards",'yard','yds','yd']:
       print("Result:" + str(number_of_inches) +" Inches= " +str(round(number_of_inches/36,2)) +" Yards")
   else:
       print("Please Enter Either Inches,Feets,or Yards.")
elif convert_from.lower() in ["feets",'ft','feet','foot']:
   number_of_feet =int(input("Enter Starting Measurements in Feets:"))
   if convert_to.lower() in ["inches",'in','inch']:
       print("Result:" + str(number_of_feet) +" Feets= " +str(round(number_of_feet*12)) +" Inches")
   elif convert_to.lower() in ["yards",'yard','yds','yd']:
       print("Result:" + str(number_of_feet) +" Feets= " +str(round(number_of_feet/3,2)) +" Yards")
   else:
       print("Please Enter Either Inches,Feets,or Yards.")
elif convert_from.lower() in ["yards",'yard','yds','yd']:
   number_of_yard =int(input("Enter Starting Measurements in Yards:"))
   if convert_to.lower() in ["inches",'in','inch']:
       print("Result:" + str(number_of_yard) +" Yards= " +str(round(number_of_yard*36)) +" Inches")
   elif convert_to.lower() in ["feets",'ft','feet','foot']:
       print("Result:" + str(number_of_yard) +" Yards= " +str(round(number_of_yard*3)) +" Feets")
   else:
       print("Please Enter Either Inches,Feets,or Yards.")
else:
    
    print("Please Enter Either Inches,Feets,or Yards.")
```
