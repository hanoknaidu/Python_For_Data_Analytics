## Default and Arbitrary Arguments

```python
def number_squared_function(number,power):
    print(number**power)
```

```python
number_squared_function(2,3)
```

```python
def number_squared_function_1(number,power=5):
    print(number**power)
number_squared_function_1(3)
```

```python
args_tuple=(5,6,7,8,9)
def args_function(*number):
    print(number[2]*number[3])
     
```

```python
args_function(*args_tuple)
```

```python
args_function(*[5,6,7,8,9])
```

```python
args_function(5,6,7,8,9)
```

```python
tuple_1=('john','jai','sam','raj')
def num_args(*word):
    print("Hello",word[2], "- Nice to meet you")
num_args(*tuple_1)
```

## Keyword And Arbitrary Arguments

```python
def number_squared_function(number,power):
    print(number**power)
```

```python
number_squared_function(4,2)
```

```python
number_squared_function(2,number=3)
```

```python
##**kwargs
def number_kwargs(**kwargs):
    print(kwargs['greetings'],kwargs['name'])
```

```python
number_kwargs(greetings='howdy',name='alex')
```

```python
def number_kwargs2(**name):
    for key,value in name.items():
        print(key + ' : '+ value)
```

```python
number_kwargs2(name='alex',name1='josh',name2='joe',name3='tina')
```

```python
my_dict = {'1': 'alex','2':'johnny','hi':'hello'}
```

```python
number_kwargs2(**my_dict)
```

```python
def wild_function(one,*two,**three):
    print(one)
    print(two[0],two[1])
    print(three.items())
```

```python
wild_function(23,*['hi','2','hello'],**my_dict)
```

## Len

```python
my_list=[1,2,3,4,8,6,9,31,'hey']
len(my_list)
```

```python
my_dict={1:'hi',2:'hello'}
len(my_dict)
```

```python
phone_numbers=['1234456689','1223344221','11212221112']
len(phone_numbers)
for num in phone_numbers:
    if len(num)==10:
        print(num)
```

```python
phone_numbers=[123-44956689,1022334421,121222177788112]
len(phone_numbers)
for num in phone_numbers:
    if len(str(num))==10:
        print(num)
```

## Format

```python
name='Alexander'
ID = "123456789"
print("My Name is" +name+  'and ID is' +ID)
```

```python
print('my name is {}. My ID is {}'. format(name,ID))
```

```python
first_name = 'Alex'
last_name = 'Alien'
print("my name is {0} {1}".format(first_name,last_name))
```

```python
name ='Alex'
adjective = 'wierd'
print("{name} is a {genius} guy.".format(name=name,genius=adjective))
```

```python
height=123.7564
print("I'm {:.1f} centimeters tall.".format(height))
```
